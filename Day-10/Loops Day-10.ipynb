```python
even_list = []
odd_list = []
for i in range(1,11):
    if i%2==0:
        even_list.append(i)
    else:
        odd_list.append(i)
print(even_list)
print(odd_list)
```

```python
#jumping statements
#break is used for stopping the iteration

for i in range(10):
    if i==4:
        break
    print(i,end=" ")
```

```python
#continue is used for skipping that particular element and continue the iteration

for i in range(10):
    if i==4:
        continue
    print(i,end=" ")
```

```python
#pass is used when we don't know what to write and to avoid the error we use pass

if 10%2==0:
    pass
```

```python
#we didn't give any instruction when i=4 we just passed it.so 4 also printed

for i in range(10):
    if i==4:
        pass
    print(i,end=" ")
print()
print(i)
```

```python
#while loop

i =1
while i<=5:
    print(i,end = " ")
    i+=1
print()
print(i)
```

```python
i = 5
while i>=1:
    print(i,end = " ")
    i-=1
print()
print(i)
```
