```python
''' 1. A user tries to withdraw money from an ATM. If withdrawal amount is less than balance,
allow it. Otherwise, show insufficient balance. Display the transaction result'''

Balance_money = 25000
Withdraw_money = int(input("Enter the amount to withdraw:"))
if Withdraw_money < Balance_money:
    Balance_money = Balance_money - Withdraw_money
    print(f'{Withdraw_money} is withdrawed')
    print(f'Balance amount in your account = {Balance_money}')
else:
    print(" Sorry for inconvience insufficient balance in your account")
```

```python
''' 2. A customer shops on an online store. If total amount is ₹2000 or more, give 20% discount.
Otherwise, no discount. Show final price message.'''

total_price = float(input("Enter the total amount:"))
discount_price = 0
if total_price >= 2000:
    discount_price = total_price * 0.20
    final_price = total_price - discount_price
    print(f'Discount price = {discount_price}')
    print(f'Final price = {final_price}')
else:
    print("You are not eligible for the discount!!!")
    print(f'Final price = {total_price}')
    
```

```python
''' 4. Read time in 24-hour format and convert it to 12-hour format with AM/PM'''

hours = int(input("Hours:"))
minutes = int(input("Minutes:"))
if hours<=12:
    print(f'Time is {hours}:{minutes} AM')
else:
    hours = hours - 12
    print(f'Time is {hours}:{minutes} PM')      
```

```python
''' 5. App suggests clothing based on temperature. Above 30 then wear Light clothes. 15–30 then
wear Normal clothes. Below 15 then wear Winter clothes.'''

temp = float(input("Enter the temperature:"))
if temp > 30:
    print("Please wear light clothes")
elif temp > 15 and temp < 30:
    print("Please wear normal clothes")
else:
    print("Please wear winter clothes")
```

```python
''' 7. Check whether the number is positive or negative or zero'''

num = int(input("Enter the number:"))
if num > 0:
    print(f'{num} is a positive number')
elif num < 0:
    print(f'{num} is a negative number')
else:
    print(f'{num} is a zero')
```
