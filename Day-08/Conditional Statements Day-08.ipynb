```python
#Checking whether the given age is eligible for voting or not using if-else

age = int(input())
if age > 18:
    print("Eligible for voting")
else:
    print("Not Eligible for voting")
```

```python
#Checking the grades based on marks using if-elif

marks = int(input("Enter the Marks:"))
if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 50:
    print("C")
elif marks >= 35:
    print("Pass")
else:
    print("Fails")
```

```python
'''If a number divisible by 3 and 5 print FizzBuzz
If a number divisible by 3 print Fizz
If a number divisible by 5 print Buzz'''

number = int(input("Enter the number:"))
if number%3==0 and number%5==0:
    print("FizzBuzz")
elif number%3==0:
    print("Fizz")
elif number%5==0:
    print("Buzz")
    
```

```python
#Checking whether the number is positive or negative using if-elif

num = int(input("Enter the number:"))
if num>0:
    print("Positive number")
elif num<0:
    print("Negative number")
else:
    print("Zero")
```

```python
#Checking whether the number is even or odd using if-else

num = int(input("Enter the number:"))
if num%2==0:
    print("Even Number")
else:
    print("Odd number")
```

```python
#Checking the greater number between two numbers using if-else

num1 = int(input())
num2 = int(input())
if num1 > num2:
    print(f'{num1} is greater')
else:
    print(f'{num2} is greater')
```

```python
#Checking the greater number among three numbers using nested-conditional statements

'''num1 = int(input("Enter the num1:"))
num2 = int(input("Enter the num2:"))
num3 = int(input("Enter the num3:"))'''

num1,num2,num3 = map(int,input("Enter 3 numbers:").split()) #giving input in single line 
if num1>num2:
    if num1>num3:
        print(f'{num1} is greater')
    else:
        print(f'{num3} is greater')
else:
    if num2>num3:
        print(f'{num2} is greater')
    else:
        print(f'{num3} is greater')
        
```

```python
#Checking the greater number among three numbers using if-elif

num1,num2,num3 = map(int,input("Enter 3 numbers:").split())
if num1>num2 and num1>num3:
    print(f'{num1} is greater')
elif num2>num1 and num2>num3:
    print(f'{num2} is greater')
elif num3>num1 and num3>num2:
    print(f'{num3} is greater')
```
