```python
#Sets
s = {1,2,3,5}
s1 = set() #set
s2 = {} #Dictionary
print(type(s),type(s1),type(s2))
```

```python
#set operations
s1={10,20,30,40}
s2={30,40,50,60}
print(s1)
print(s2)

#union
print(s1.union(s2))
print(s1|s2)
```

```python
#intersection
print(s1.intersection(s2))
print(s1&s2)
```

```python
#difference
print(s1.difference(s2))
print(s1-s2)
print(s2.difference(s1))
print(s2-s1)
```

```python
#symmetric_difference
print(s1.symmetric_difference(s2))
print(s1^s2)
```

```python
s3 = {26,45,18,1,4}
s4 = {26}
#issuperset()
print(s3.issuperset(s4))
#issubset()
print(s3.issubset(s4))
print(s4.issubset(s3))
```

```python
#Dictionaries

d = {}
d1 = dict()
d2 = {'a':1,
      'b':2,
      'c':3}
print(type(d),type(d1),type(d2))
```

```python
#checking the value of a particular key
print(d2['a'])
#getting all the values in dict
print(d2.values())
#fetting all the keys in dict
print(d2.keys())
#getting the keys and values in dict
print(d2.items())
```

```python
#reassigning c value
d2['c'] = 6
print(d2)
```

```python
#adding a new key and value
d2['d'] = 4
print(d2)
```

```python
d2.update({'e':5})
print(d2)
d2.update(f=6)
print(d2)
```

```python
#getting the value of a key 
print(d2.get('c'))
```

```python
#trying to get the value which doens't have
print(d2['z'])
```

```python
#trying to get the value which doesn't have in dict without error
print(d2.get('z',0))
print(d2.get('a',0))
```

```python
#popping the item
d2.pop('b')
print(d2)
```

```python
#popping the item which is not present in dict
d2.pop('b')
print(d2)
```

```python
#popping the item which is not present in dict without getting error
d2.pop('b',None)
print(d2)
```

```python
#popping or removing the last item
d2.popitem()
print(d2)
```

```python
#clearing all the items from dict
d2.clear()
print(d2)
```

```python
#deleting the entire dictonary
del d2
print(d2)
```
