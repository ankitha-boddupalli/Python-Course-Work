```python
#numeric datatype

number = 100
percentage = 9.56
c = 2+3j
c2 = complex(number)
print(c2,type(c2))
print(number,type(number))
print(percentage,type(percentage))
print(c,type(c))
```

```python
#sequence datatype

s = "Ankitha"
lst = [10,20,30]
t = (1,2,3,4)
t2 = {100,}
print(t2,type(t2))
print(type(s),type(lst),type(t))
```

```python
#Boolean datatype
#True/False

boole1 = True
boole2 = False
print(type(boole1))
print(boole1 + boole2)
```

```python
#mapping datatype - dictionary

d = {"name":"Codegnan",
  "location":"Hyderabad",
  "branch":4}
print(d, type(d))
```

```python
#sets datatype

s = {1,2,3,4}
print(s,type(s))
```

```python
#Empty datatypes generation

lst1 = [] #list()
t1 = () #tuple()
str1 = "" #str()
d1 = {} #dict()
s1 = set() #set()
print(lst1,type(lst1))
print(t1,type(t1))
print(str1,type(str1))
print(d1,type(d1))
print(s1,type(s1))
print(type(str()))
```

```python
#Implicit Typecasting

num1 = 100
num2 = 6.78
num3 = num1+num2
print(type(num1),type(num2))
print(num3,type(num3))
```

```python
#Explicit Typecasting

n = input()
print(type(n))
n2 = int(n)
print(type(n2))
```

```python
#Type conversions

str1 = "ankitha"
print(str1)
lst2 = list(str1)
print(lst2)
t2 = tuple(lst2)
print(t2)
set2 = set(lst2)
print(set2)
```

```python
#taking input from user

num = int(input("Enter the number:")) #for integer
num1,num2,num3 = map(int,input("Enter three numbers:").split()) #three integers in a row
s = input("Enter the string:") #string
list1 = list(map(int,input("Enter the list:").split(","))) #list
tuple1 = tuple(map(int,input("Enter the tuple:").split())) #tuple
set1 = set(map(int,input("Enter the set:").split())) #set
print(f'Entered number = {num}')
print(f'Entered three numbers = {num1} {num2} {num3}')
print(f'Entered string = {s}') 
print(f'Entered list = {list1}')
print(f'Entered Tuple ={tuple1}')
print(f'Entered Set = {set1}')
```
