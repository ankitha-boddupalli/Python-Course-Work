```python
# right angled triangle (second method)

row = int(input("Enter no of rows:"))
column = int(input("Enter no  of columns:"))
for i in range(1,row+1):
    for j in range(1,column+1):
        if i>=j:
            print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```

```python
# Right-angled triangle with numbers

row = int(input("Enter no of rows:"))
column = int(input("Enter no  of columns:"))
n = 1
for i in range(1,row+1):
    for j in range(1,column+1):
        if i>=j:
            print(n,end = " ")
            n += 1
        else:
            print(" ",end = " ")
    print()
```

```python
#Inverted right angled triangle

n = 4
for i in range(n):
    for j in range(n-i):
        print("*",end = " ")
    print()
```

```python
n = int(input("Enter value of n:"))
for i in range(n):
    for j in range(n-i):
        if (i+j) <= (n-1):
            print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```

```python
#pyramid

n = 3
for i in range(n):
    # for spaces
    for j in range(n-i-1):
        print(" ",end = " ")
    #for stars
    for j in range(2*i+1):
        print("*",end = " ")
    print()
```

```python
#invert-pyramid

n = 3
for i in range(n):
    #spaces
    for j in range(i):
        print(" ",end=" ")
    #stars
    for j in range(2*n-(2*i+1)):
        print("*",end=" ")
    print()
```

```python
#rhombus

n = 4
for i in range(n):
    # for spaces
    for j in range(n-i-1):
        print(" ",end = " ")
    #for stars
    for j in range(2*i+1):
        print("*",end = " ")
    print()
n=n-1
for i in range(n):
    #spaces
    for j in range(i+1):
        print(" ",end=" ")
    #stars
    for j in range(2*n-(2*i+1)):
        print("*",end=" ")
    print()
```
