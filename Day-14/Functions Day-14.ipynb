```python
#counting the voewls and consonants in a given string using function

def alphabet_count(str1):
    vowels_count = 0
    consonant_count = 0
    for ch in str1:
        if ch.isalpha():
            if ch in "aeiouAEIOU":
                vowels_count += 1
            else:
                consonant_count += 1
    return [vowels_count,consonant_count]
            
```

```python
res = alphabet_count("Python Project")
print(res)
```
