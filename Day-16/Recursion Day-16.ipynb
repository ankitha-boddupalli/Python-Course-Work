```python
# Pass by object referrence - immutable data types
def f(x):
    print(id(x))
    x = x+10
    print(id(x))
    return x
```

```python
a = 100
print(id(a))
f(a)
```

```python
# Pass by object referrence - mutable data types
def f1(x):
    print(id(x))
    x.append(100)
    print(id(x))
    return x
    
```

```python
b = [10,20,30,40]
print(id(b))
f1(b)
```

```python
# Pass by object referrence - mutable data types(re-assigning)
def f1(x):
    print(id(x))
    x = [10,20,30,40]
    print(id(x))
    return x
    
```

```python
b = [10,20,30,40]
print(id(b))
f1(b)
```

```python
# Pass by object referrence - immutable data types(re-assiging)
def f(x):
    print(id(x))
    x = 100
    print(id(x))
    return x
```

```python
a = 100
print(id(a))
f(a)
```

```python
# Recursion
#factorial of a number using recursion

def factorial(n):
    # base condition
    if n==1 or n==0:
        return 1
    # recursive function call
    return n*factorial(n-1)
```

```python
factorial(5)
```

```python
# add 'N' natural numbers using recursion

def n_natural_sum(n):
    if n==1 :
        return 1
    return n+n_natural_sum(n-1)
```

```python
n_natural_sum(10)
```

```python
# add only even numbers

def even_natural_sum(n):
    if n==2:
        return 2
    return n + even_natural_sum(n-2)
```

```python
even_natural_sum(6)
```

```python
# if even add even natural numbers, if odd add odd natural numbers

def n_natural_sum(n):
    if n%2==0:
        if n==2:
            return 2
        return n + n_natural_sum(n-2)
    else:
        if n==1:
            return 1
        return n + n_natural_sum(n-2)
        
```

```python
n_natural_sum(10)
```
