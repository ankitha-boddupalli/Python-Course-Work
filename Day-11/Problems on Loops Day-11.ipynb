```python
# factors

num = int(input("Enter a number:"))
for factor in range(1, num+1):
    if num % factor == 0:
        print(f'Factor of {num} is {factor}')
```

```python
# count of odd numbers within a range

num = int(input("Enter the number: "))
count = 0
for i in range (1,num+1):
    if i % 2 != 0:
        count = count + 1
print(f'count of odd numbers: {count}')
```

```python
# prime number

number = int(input("Enter the number:"))
count = 0
for i in range(1,number+1):
    if number % i == 0:
        count = count+1

if count == 2:
    print("Prime Number")
else:
    print("Not a Prime")
```

```python
# Perfect number or not

num = int(input("Enter the number:"))
summ = 0
for i in range (1,num):
    if num % i == 0:
        summ = summ + i
        print(f'factor of {num} is {i}')
        
print(f'sum={summ}')
if summ == num:
    print("Perfect number")
else:
    print("Not perfect number")
        
```

```python
# Find the minimum number in a list

lst = [10, 28,43,7,90,4,78,-3,-8,56]
mini = lst[0]
for num in lst:
    if num < mini:
        mini = num
print(f'minimum value = {mini}')
        
```

```python
# Find the maximum number in a list

lst = [10, 28,43,7,90,4,78,-3,-8,56]
maxi = lst[0]
for num in lst:
    if num > maxi:
        maxi = num
print(f'maximum value = {maxi}')
```

```python
# sum of elements in a list

lst = [1,2,3,4,5]
summ = 0
for num in lst:
    summ = summ + num
print(f'sum of elements in list= {summ}')
```
