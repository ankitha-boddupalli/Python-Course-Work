```python
#square shaped stars

rows = int(input("Enter the no of rows:"))
columns = int(input("Enter the no of columns:"))

for i in range(rows):
    for j in range(columns):
        print("*",end = " ")
    print()
```

```python
#Hallow shaped square

rows = int(input("Enter the no of rows:"))
columns = int(input("Enter the no of columns:"))

for i in range(rows):
    for j in range(columns):
        if i==0 or  i==rows-1 or j==0 or j==columns-1:
            print("*",end=" ")
        else:
            print(" ",end = " ")
    print()
    
```

```python
#Diagonals of a square

rows = int(input("Enter the no of rows:"))
columns = int(input("Enter the no of columns:"))

for i in range(rows):
    for j in range(columns):
        if i==j or (i+j)==(rows-1):
            print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```
