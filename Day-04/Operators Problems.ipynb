```python
'''1. Area & perimeter calculation of a house'''
len=int(input())
width=int(input())
area=len*width
perimeter=2*(len+width)
print(f'Area of the house is {area}, Perimeter of the house is {perimeter}')
```

```python
'''2. Find out the Volume of Water Tank'''

r=int(input("Enter Radius:"))
h=int(input("Enter Height: "))
v=3.14*r*r*h
print(f'Volume of the Water Tank {v}')
```

```python
'''3. You are developing a small application for a walk-time app. In the app, users usually enter
their walking time in N minutes. Your task is to convert the minutes and return an integer
representing the equivalent value in seconds'''

minutes=int(input("Enter no. of minutes:"))
seconds=minutes*60
print(f'In seconds {seconds}')
```

```python
'''4. A student plans to read T pages of a book. On the first day, she reads p1 pages, and on the
second day, she reads p2 pages. Write an expression to find how many pages she still needs to
read to complete the book.'''

T=int(input("Enter total no. of pages:"))
p1=int(input("Enter pages completed on day 1:"))
p2=int(input("Enter pages completed on day 2:"))
R=T-(p1+p2)
print(f'Remaining Pages:{R}')
```

```python
'''5. Find the Square of a given number'''

N=int(input("Enter the Number:"))
S=N*N
print(f'Square of the number:{S}')
```

```python
'''6. You have a daily data pack of D GB. You used u1 GB in the morning and u2 GB in the
afternoon. Write an expression to find how much data is still remaining for the day.'''

D=int(input("Enter Daily Data Pack:"))
U1=int(input("Enter morning used data:"))
U2=int(input("Enter afternoon used data:"))
R=D-(U1+U2)
print(f'Remaining Data:{R}')
```

```python
'''7. You bought 3 items costing a, b, and c. Find the total amount to pay.'''

a=int(input("Enter the cost of first item:"))
b=int(input("Enter the cost of second item:"))
c=int(input("Enter the cost of third item:"))
total=a+b+c
print(f'Total amount to pay:{total}')
```

```python
'''8. Without using built-in functions, find the last digit of 4 digit number'''

n=int(input("Enter a 4-digit number:"))
r=n%10
print(f'last digit is:{r}')
```

```python
'''9. Find the average of three numbers'''

n1=int(input("Enter the first number:"))
n2=int(input("Enter the second number:"))
n3=int(input("Enter the third number:"))
average=(n1+n2+n3)//3
print(f'Average:{average}')
```
