```python
#shallow copy

import copy
l1=[1,2,3,[10,20]]
l2=copy.copy(l1)
print(l1)
print(l2)
l1[0]=100
print(l1)
print(l2)
l1[3][0]=1000
print(l1)
print(l2)
```

```python
#Deep copy

import copy
l1=[1,2,3,[10,20]]
l2=copy.deepcopy(l1)
print(l1)
print(l2)
l1[0]=100
print(l1)
print(l2)
l1[3][0]=1000
print(l1)
print(l2)
```
